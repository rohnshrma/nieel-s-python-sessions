# Collection
used to store multiple values of similar or different data types

- `list` : ordered, indexed, mutable (changable) , allows duplicates
- `tuple` : ordered, indexed, immutable(unchangable) , allows duplicates
- `set` : unordered, unindexed, immutable, no duplicates allowed
- `dictionary` : ordered, unindexed, changable, no duplicates allowed

## List
ordered, indexed, mutable (changable) , allows duplicates

`[ ]` are used to create a list where items are separated using `,`

```python
names = ["John","Josie","James" , "Jackie", "Jimmy" , "Jackson"]
marks = [23,23,34,2,3,35,46,43,434,3,323,343]
```

```python
# type() function returns the data type of the value / variable specified
```

```python
print(type(names) , type(marks))
```

```python
print(len(names) , len(marks))
```

```python
print(marks) # ordered and allows duplicates
```

```python
print(marks[2]) # indexed
```

```python
marks[2] = 100 # mutable
print(marks)
```

```python
# add a new item at the end of the list
marks.append(1010101)
print(marks)
```

```python
# insert new item at the specified index
marks.insert(1 , 3434343)# position , item
print(marks)
```

```python
marks.extend([3567, 78787, 68668])
print(marks)
```

```python
# removes and returns items on -1 index by default, or item on the index specified
marks.pop() # -1 : 68668
marks.pop(1) # 1 : 3434343
print(marks)
```

```python
# removes the item specified, raises an error if not found
marks.remove(100)
```

```python
print(marks)
```

```python
# returns number of occurances of the item specified
marks.count(3)
```

```python
# returns the index of the first occurance of the item specified
marks.index(3)
```

```python
# sorts the items (a-z), or (z-a) is specified
marks.sort()
print(marks)
marks.sort(reverse=True)
print(marks)
```

```python
marks[ : : -1] # returns a new reversed list 
```

```python
marks.reverse()
```

```python
print(marks)
```

```python
# returns a copy of list
x = marks.copy()
print(x)
```

```python
# empties a list
x.clear()
print(x)
```

```python
del x
```
