# set

A set is a collection that keeps only **unique** items. Think of it like a bag of marbles, you can't tell which one is "first", and you can't have two identical ones.

- created with `{ }` and items separated by a comma `,`
- **unordered** -> no fixed position, order can change every time you print
- **unindexed** -> no `set[0]`, since there is no position to point at
- **no duplicates** -> repeated items are dropped automatically
- the set itself **is changeable** (you can add and remove items), but the items inside must be unchangeable (numbers, strings, tuples). A list can't go inside a set.
- an empty `{}` is a **dictionary**, not a set. For an empty set use `set()`

## creating a set and its properties

```python
# curly brackets make a set, repeated numbers (12 and 56) will be kept only once
marks = {12,23,12,12,12,123,45,56,34,56,55,64,767,3}
```

```python
# type() tells the data type, len() counts the items AFTER duplicates are removed
print(type(marks) , len(marks))
```

```python
# print the whole set, notice no duplicates and the order is not what we typed
print(marks)  # unordered and no duplicates
```

```python
# trying to read by position fails because a set has no index
print(marks[0]) # unindexed -> TypeError
```

```python
# trying to change by position also fails for the same reason
marks[0] = 10000 # no index, so no item assignment -> TypeError
```

## adding and removing items

| method | what it does | if item is missing |
|---|---|---|
| `add(x)` | puts one new item in | -- |
| `update(iterable)` | puts many items in at once | -- |
| `pop()` | removes a **random** item | error if set is empty |
| `remove(x)` | removes that item | **KeyError** |
| `discard(x)` | removes that item | does nothing, no error |
| `clear()` | empties the set | -- |

```python
# add() puts one new item in the set
marks.add(34345)
# print to see the new item is there (at some random spot, sets have no order)
print(marks)
```

```python
# pop() removes one item, but we can't choose which one, python picks
marks.pop()
# print to check that one item is gone
print(marks)
```

```python
# remove() deletes the item whose value we give
marks.remove(3)
# 555555 is not in the set, so remove() throws a KeyError and the next line never runs
marks.remove(555555)
print(marks)
```

```python
# discard() also deletes by value
marks.discard(3)
# 555555 is not there, but discard() just ignores it, no error
marks.discard(555555)
# print the final set
print(marks)
```

## maths on sets (like venn diagrams)

Two sets `x` and `y`. Every operation has two versions:
- plain version (`intersection`) -> gives a **new** set, `x` stays same
- `_update` version (`intersection_update`) -> **changes x itself** and returns nothing

| method | meaning | venn part |
|---|---|---|
| `union` | everything from both | whole picture |
| `intersection` | only what is common | the overlap |
| `difference` | in x but not in y | x without overlap |
| `symmetric_difference` | in one of them but not both | both sides without overlap |

```python
# two sets, 4 and 5 are in both
x = {1,2,3,4,5}
y = {4,5,6,7,8}
```

```python
# new set with the common items -> {4, 5}, x is untouched
x.intersection(y)
```

```python
# same thing but x itself becomes {4, 5}
x.intersection_update(y)
```

```python
# items that are in x but not in y
# (x was changed by the cell above, so the answer here depends on that. Re-run the x = {1,2,3,4,5} cell to get {1, 2, 3})
x.difference(y)
```

```python
# x is changed so it keeps only the items that were not in y
x.difference_update(y)
```

```python
# items which are not common, i.e. present in only one of the two sets
x.symmetric_difference(y)
```

```python
# x becomes the uncommon items of both
x.symmetric_difference_update(y)
```

```python
# joins both sets into one, common items still appear once
x.union(y)
```

## comparing sets

- `issuperset` -> does x contain **all** of y?
- `issubset` -> is every item of y found inside x?
- `isdisjoint` -> do they have **nothing** in common? (True = no overlap at all)

```python
# x is bigger, y is a small part of it
x = {1,2,3,4,5,6,7,8}
y = {6,7,8}
```

```python
# x has everything y has -> True
x.issuperset(y)
```

```python
# all of y's items are inside x -> True
y.issubset(x)
```

```python
# a set with strings, nothing like x
names = {"a","b","c"}
```

```python
# no common item between names and x -> True
names.isdisjoint(x)
```

```python
# update() adds many items together, here it takes another set
names.update({120,123})
```

```python
# just typing the variable name shows its value
names
```

### small correction to my earlier note
Earlier I wrote that a set is "immutable". That is not fully right. A set **can** be changed (`add`, `remove`, `update` all work). What we can't do is change an item by index, because there is no index.
