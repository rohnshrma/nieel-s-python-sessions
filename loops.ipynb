# loops: the `while` loop

A loop **repeats** a block of code again and again, so we don't have to write the same line 100 times. Imagine printing numbers from 1 to 100, with a loop it is just 4 lines.

The `while` loop keeps running **as long as its condition is True**. The moment the condition becomes False, the loop stops and python moves on to the line after it.

## fixed iteration
When we already know how many times the loop should run (print 1 to 10, print 20 to 40...), every while loop needs three parts:

| part | what it does |
|---|---|
| **initialization** | a starting value for the counter variable, written before the loop |
| **condition** | checked before every round, if it is True the loop body runs |
| **update** (increment / decrement) | changes the counter inside the loop so that the condition will eventually become False |

```
initialization
while condition:
    code to be executed
    update
```
- the loop body is the indented part, just like in `if`
- **increment** = counter goes up (`i += 1`), **decrement** = counter goes down (`i -= 1`)
- one full run of the body is called one **iteration**

## 1. counting up

````python
i = 1 # initialization, the counter starts at 1
while i <= 10: # condition, keep going while i is 10 or less
    print(i) # code to be executed, prints the current value of i
    i+=1 # update, add 1 to i, without this the loop would never end
# when i becomes 11, the condition 11 <= 10 is False and the loop stops
````

## 2. different step size
The update does not have to be `+1`. Adding 2 every time skips every other number.

````python
i = 20 # start from 20
while i<= 40: # run until i crosses 40
    print(i) # print the current number
    i += 2 # jump by 2 each time -> 20, 22, 24 ... 40
````

## 3. counting down (decrement)
Start high, go low. The condition changes direction too (`>=`) and the update subtracts.

````python
i = 100 # start at the top

while i >= 1: # keep going while i is still 1 or more
    print(i) # print the current number
    i-=1 # subtract 1 each time, so the counter moves downwards
````

## 4. the position of the update matters
Same code as the first example, but the **update is above the print**. Because `i` is changed first, the first number printed is 2 and the last is 11, even though the condition says `i <= 10`.

Why 11 appears: when `i` is 10 the condition is True, the body starts, `i` becomes 11 and 11 gets printed. Only after that the condition is checked again, and now it is False.

````python
i = 1 # initialization
while i <= 10: # condition
    i+=1 # update happens FIRST, so i is already 2 when it gets printed
    print(i) # code to be executed -> prints 2 to 11
````

## 5. infinite loop (careful!)
If the update is forgotten (or moves the counter in the wrong direction) the condition never becomes False and the loop **never stops**. In Jupyter press the stop button (interrupt the kernel) to escape.

```
i = 1
while i <= 10:
    print(i)     # no i += 1, so i is always 1 and the loop runs forever
```
This is shown as text and not as code on purpose, so it doesn't freeze the notebook.

## 6. using a loop with data we already know

Counter variables are also good for walking through a list by its **index**, because index numbers are just a count from 0. `len()` tells us how many items exist, so the loop should stop **before** `len(list)` (the last index is `len - 1`).

````python
fruits = ["apple", "banana", "mango", "grapes"]
i = 0 # index of the first item is 0
while i < len(fruits): # len is 4, valid indexes are 0 to 3, so we use < and not <=
    print(i, fruits[i]) # prints the position and the item at that position
    i += 1 # move on to the next index
````

## 7. keeping a running total
A variable made **outside** the loop can collect something in every round. This pattern (total / count) is used in most real programs.

````python
total = 0 # starts empty, created before the loop so it isn't reset every round
i = 1
while i <= 5: # runs for 1, 2, 3, 4, 5
    total += i # adds the current number to the running total
    i += 1 # update
# after the loop ends the total is 1+2+3+4+5
print(total)
````
