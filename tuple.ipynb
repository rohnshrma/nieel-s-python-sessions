# tuple

A tuple is like a list that is **locked**. Once made, you can't change it. Good for data that should not change, like days of a week or a (latitude, longitude) point.

- **ordered** -> items stay in the position you put them
- **indexed** -> `t[0]` gives the first item (counting starts from 0)
- **unchangeable (immutable)** -> no add, no edit, no remove
- **allows duplicates**

---
- `( )` can be used to create a tuple
- but `( )` alone does not make a tuple, the **comma** does
- putting multiple values in a variable separated by comma, with or without `( )`, makes it a tuple
- for a tuple with a single item put a comma `,` after it (with or without `( )`)

```python
# only brackets, no comma -> python reads it as a plain number, so it is an int
x = (123)
# commas between values -> a tuple, even without brackets
y =  123,456,678
# a single value followed by a comma -> a tuple with one item
z = 123,
# check the type of each one
print( type (  x ))   # int
print( type (  y ))   # tuple
print( type (  z ))   # tuple
```

## indexing and immutability

```python
# tuple with duplicates (34 twice, 4 twice), brackets are not needed
marks = 34,56,34,4,534,4,5
```

```python
# type and the content together
print(type(marks) , marks)
```

```python
# order is same as typed and duplicates are kept
print(marks) # ordered and allows duplicates
```

```python
# index 1 is the second item -> 56
print(marks[1]) # indexed
```

```python
# trying to change an item -> TypeError, tuples are locked
marks[1] = 1000 # immutable (unchangable)
```

## tuple methods
Tuples have only two methods because you can't change them.
- `count(x)` -> how many times x appears
- `index(x)` -> position of the **first** x

```python
# 4 appears two times in marks -> 2
marks.count(4)
```

```python
# first 4 is at index 3 -> 3 (the second 4 is ignored)
marks.index(4)
```

## unpacking
Taking the items of a tuple out into separate variables. The number of variables must match the number of items, unless you use `*`.
- `*name` collects the left over items into a **list**

```python
# a tuple of 3 marks
yourMarks = 98 ,56,78
```

```python
# reading each item by its index, one by one
print(yourMarks[0])
print(yourMarks[1])
print(yourMarks[2])
```

```python
# unpacking: 3 items go into 3 variables in order, x=98, y=56, z=78
x, y ,z = yourMarks
```

```python
# check that every variable got its value
print(x)
print(y)
print(z)
```

```python
# star unpacking: b takes the last item, a (with *) takes everything before it as a list
*a , b = yourMarks
```

```python
# a = [98, 56] (a list!), b = 78
print(a , b)
```

## the trick to "change" a tuple
We can't edit it, but we can **convert -> edit -> convert back**:
1. `list(tuple)` makes an editable copy
2. change the list
3. `tuple(list)` locks it again

Note this creates a *new* tuple, the old one is not modified.

```python
# step 1: tuple -> list
marksList = list(yourMarks)
# step 2: add three new marks at the end, then replace the first one
marksList.extend([10000,199999,44444444444])
marksList[0] = 999999
# step 3: list -> tuple and store it back in yourMarks
yourMarks = tuple(marksList)
# see the final result
print(yourMarks)
```
