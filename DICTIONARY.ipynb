# dictionary

A dictionary stores data as **key : value** pairs, just like a real dictionary where the word is the key and the meaning is the value. You don't look things up by position, you look them up by name.

- created with `{ }`, each item written as `key : value`, items separated by a comma `,`
- **ordered** -> items stay in the order they were added (python 3.7+)
- **unindexed** -> no `d[0]`, we use the **key** to get the value
- **changeable (mutable)** -> add, edit and remove items any time
- **no duplicate keys** -> if a key is repeated, the last value wins. Values can repeat freely
- keys must be unchangeable (string, number, tuple). Values can be anything, even a list or another dictionary
- an empty dictionary is `{}`

## creating a dictionary and its properties

```python
# one dictionary describing a car
car = {
    "model" : "figo",                       # key "model", value "figo"
    "brand" : "ford",                       # key "brand", value "ford"
    "year"  : 2020,                         # first "year" entry
    "year"  : 2024,                         # same key again -> this value replaces 2020
    "colors" : ["red" , "grey" , "black"]   # a value can be a list
}
```

```python
# type() confirms it is a dict
print(type(car))
```

```python
# "year" appears only once with 2024, so len is 4 and not 5
print(car , len(car)) # ordered and no duplicate keys allowed
```

```python
# the key goes inside [ ] and gives back its value -> figo
print(car["model"]) # unindexed (keys are used to access values)
```

## changing and adding items

```python
# key already exists, so its value is replaced
car["model"] = "mustang" # mutable
# print to see the new model
print(car)
```

```python
# key does not exist yet, so a brand new pair is added at the end
car["isAvailable"] = False
print(car)
```

## reading values: `[ ]` vs `get()`

| way | key found | key missing |
|---|---|---|
| `d["key"]` | returns value | **KeyError** (program stops) |
| `d.get("key")` | returns value | returns `None`, no error |
| `d.get("key", default)` | returns value | returns your default |

```python
# square brackets, key exists -> mustang
car["model"] # get the value from a key, raises a KeyError if key not found
```

```python
# get() with an existing key -> mustang
car.get("model") # get the value from a key, returns None if key not found
```

```python
# "mileage" is not in car yet, so this crashes with KeyError
car["mileage"]
```

```python
# same missing key but get() stays calm and gives None (nothing is shown in the output)
car.get("mileage")
```

## keys, values and items
These return **views**, a live window over the dictionary. They look like lists but are not exactly lists, wrap with `list()` if you need a real list.

```python
# all the values only
car.values()
```

```python
# all the keys only
car.keys()
```

```python
# every pair as a tuple (key, value)
car.items()
```

## removing items
- `pop(key)` -> removes that key and **returns its value**
- `popitem()` -> removes the **last added** pair and returns it as a tuple
- `del d[key]` -> removes that key (returns nothing)
- `clear()` -> empties the whole dictionary

```python
# removes the "isAvailable" pair and hands back its value -> False
car.pop("isAvailable")
```

```python
# removes the last pair in the dictionary ("colors") and returns it as a tuple
car.popitem()
```

## update
`update()` takes another dictionary and merges it in:
- key already there -> value is **replaced**
- key not there -> pair is **added**

```python
# "year" exists so it becomes 2019, "mileage" is new so it gets added
car.update({"year" : 2019 , "mileage" : "20kmpl"})
```

```python
# check the merged result
print(car)
```

## fromkeys and setdefault

- `dict.fromkeys(keys, value)` -> builds a new dictionary where every key gets the **same** value
- `setdefault(key, value)` -> if key exists, just returns its value and changes nothing. If it doesn't exist, adds it with the value (or `None` if no value is given) and returns it

```python
# make a dictionary from a list of keys, all of them start with the value 0
x = dict.fromkeys(["nieel" , "rohan" , "arpita"] , 0)
```

```python
# view the new dictionary -> {'nieel': 0, 'rohan': 0, 'arpita': 0}
x
```

```python
# "nieel" already exists, so nothing changes and it returns the old value 0 (not 100)
x.setdefault("nieel" , 100)
```

```python
# "akbar" is new and no value is given, so it is added with None
x.setdefault("akbar" , )
```
