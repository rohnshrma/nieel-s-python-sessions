# control flow statements (cfs): conditionals

Till now python ran our code line by line, top to bottom, every single line. Conditionals let the program **take a decision**: run some code only *if* something is true, and skip it otherwise. Just like in real life, "if it rains, take an umbrella, else wear sunglasses".

- the condition is anything that becomes `True` or `False` (comparison and logical operators from the last notebook)
- a colon `:` goes at the end of the `if` / `elif` / `else` line
- the code that belongs to the condition is pushed in by **indentation** (4 spaces or one tab). Indentation is not decoration in python, it is how python knows which lines are inside the `if`
- only **one** block runs in an if / elif / else chain, the first one whose condition is true

| keyword | meaning |
|---|---|
| `if` | check the first condition |
| `elif` | "else if", check another condition only if all the ones above were false. We can have many |
| `else` | no condition, runs when nothing above was true. Optional, and always last |

## 1. the `if` statement

```
if condition:
    code to be executed
```

````python
# 12 < 10 is False, so python skips the indented line, nothing is printed at all
# (the message says "Greater" but the check is for less than, so it never shows up)
if 12 < 10:
    print("Greater")
````

````python
# this time the condition is True, so the indented line runs
if 12 > 10:
    print("12 is greater than 10")
# this line is NOT indented, so it is outside the if and runs every time
print("this always prints")
````

## 2. `if` ... `else`
`else` is the plan B, it runs when the `if` condition is false. Exactly one of the two blocks always runs.

````python
# condition is False (12 is not less than 10), so python jumps to else
if 12 < 10:
    print("Greater")    # skipped
else:
    print("lesser")     # runs -> lesser
````

## 3. `if` ... `elif` ... `else`
When there are more than two possibilities. Python checks from the top and stops at the first true condition.

`input()` asks the user to type something. It always gives back a **string**, so `int()` is used to turn it into a number before we compare.

````python
# input() shows the message, waits for the user to type, and returns the typed text
# int() converts that text into a whole number, so we can compare it
n = int(input("Enter a number"))
# first check: is it above zero
if n > 0:
    print("Positive")
# only checked if n > 0 was false: is it below zero
elif n < 0:
    print("Negative")
# if both checks failed, the only possibility left is zero
else:
    print("Zero")
````

## 4. joining conditions with `and` / `or` / `not`
We can use the logical operators inside an `if` to check more than one thing together.

````python
# two variables to test
age = 20
has_id = True
# both must be True for the person to enter
if age >= 18 and has_id:
    print("entry allowed")
else:
    print("entry denied")
````

## 5. nested `if`
An `if` inside another `if`. The inner one is checked only when the outer one is true. Each level needs its own extra indentation.

````python
# first check is the username, only if that is right we check the password
username = "rohan"
password = "1234"
if username == "rohan":
    # we are inside the first if, so this runs only for the right username
    if password == "1234":
        print("login successful")
    else:
        print("wrong password")
else:
    print("user not found")
````

## 6. why the order of `elif` matters
Python stops at the **first** true condition. Always put the most strict condition first.

````python
# marks 95 is greater than 90 AND greater than 75, but only the first true block runs
marks = 95
if marks >= 90:
    print("grade A")        # this runs, python stops here
elif marks >= 75:
    print("grade B")        # never checked
elif marks >= 50:
    print("grade C")        # never checked
else:
    print("fail")
````

## 7. one line `if else` (short form)
```
value_if_true if condition else value_if_false
```
Good for small decisions that just choose between two values.

````python
# the result is stored in a variable instead of printing inside the if
num = 7
result = "even" if num % 2 == 0 else "odd"
# 7 % 2 is 1, so the condition is False and "odd" is chosen
print(result)
````
