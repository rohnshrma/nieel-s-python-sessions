# indexing 
counting starting from 0 , used to access characters of a string

---
[ ] are used to use indexing

```python
# h e l l o
# 1 2 3 4 5 : normal counting (getting length)
# 0 1 2 3 4 : positive indexing (L -> R)
#-5-4-3-2-1 : negative indexing (R -> L)
```

---

- the first character is always on the 0th position (+)
- the last character is always on the `len(str) - 1` = th position  (+)
- the first character is always on the -1th position (-)

str[start : stop : step]

````python
x = "hello world this is john doe"

print(x[2])                 # l -> 3rd character (counting starts from 0)
print(x[6])                 # w
print(x[10])                # d
print(x[ len(x) - 1])       # e -> last character using length - 1
print(x[-1])                # e -> -1 is a shortcut for the last character
print(x[-8])                # j -> 8th character from the back
````

````python
# slicing : x[start : stop] -> start is included, stop is NOT included
print(x[0 : 4]) # 0-3
print(x[0 : 5]) # 0-4
print(x[20 : 28]) # 20 - 27
print(x[20 : ]) # 20 - end (stop left empty = go till the end)
print(x[ : 11]) # start - 10 (start left empty = begin from 0)
print(x[ -8 : ]) # -8 : end (negative numbers work in slicing too)
````

````python
# strings are immutable, so every method gives a NEW string, original x stays same
# return a new string having all characters in uppercase
print(x.upper())
# return a new string having all characters in lowercase
print(x.lower())
# return a new string having first character of each word in upper and rest in lowercase
print(x.title())
# return a new string having first character of entire string in upper and rest in lowercase
print(x.capitalize())
# small becomes capital and capital becomes small
print("HEllO WOrLd".swapcase())
````

````python
# replace() swaps every old piece of text with a new one
print(x.replace( "o" ,"x" )) # old,  new
print(x.replace( "o" ,"x",count=2)) # old , new, limit (only first 2 are changed)
````

````python
# strip family removes extra spaces (or any given characters) from the edges only
print("              hello world         ".strip())    # both sides
print("              hello world         ".lstrip())   # left side only
print("              hello world         ".rstrip())   # right side only
````

# Remaining string methods
Now the rest of the methods. Remember : a string can never be changed once made (immutable), so every method returns a **new** string and the original stays as it is.

## Searching : find, rfind, index, rindex
- `find()` gives the position of the first match, `-1` if not found
- `rfind()` same thing but searches from the right side
- `index()` / `rindex()` work like find, but throw `ValueError` if not found

````python
x = "hello world this is john doe"

print(x.find("o"))          # 4  -> first 'o' is at index 4
print(x.find("o", 5))       # 7  -> start searching from index 5
print(x.find("xyz"))        # -1 -> not found, no error
print(x.rfind("o"))         # 26 -> last 'o', searched from the right
print(x.index("world"))     # 6  -> same as find when the text is present
print(x.rindex("is"))       # 17 -> last "is" (inside "this is")

try:
    x.index("xyz")          # index() does not return -1, it raises an error
except ValueError as e:
    print("Error :", e)
````

## Splitting and joining : split, rsplit, splitlines, partition, rpartition, join
- `split()` breaks a string into a **list** of pieces
- `join()` does the opposite, sticks a list of strings together using a separator

````python
print(x.split())            # no argument -> splits on any whitespace
print(x.split("o"))         # splits wherever 'o' comes
print(x.split(" ", 2))      # maxsplit = 2 -> only first 2 splits happen
print(x.rsplit(" ", 2))     # same, but splitting starts from the right

text = "line one\nline two\nline three"
print(text.splitlines())    # breaks at every line break
print(text.splitlines(True))  # True -> keeps the \n at the end of each line

print(x.partition("is"))    # (before, separator, after) -> splits only at the FIRST match
print(x.rpartition("is"))   # same, but at the LAST match
print(x.partition("xyz"))   # not found -> (whole string, '', '')

words = ["python", "is", "easy"]
print(" ".join(words))      # python is easy  -> separator goes between items
print("-".join(words))      # python-is-easy
print("".join(["a", "b"]))  # ab
````

## Alignment and padding : center, ljust, rjust, zfill, expandtabs
Used for neat looking output. First value is total width, second (optional) is the filler character.

````python
print("hi".center(10, "*"))     # ****hi****  -> text in the middle
print("hi".ljust(10, "-") + "|") # hi--------|  -> text on the left
print("hi".rjust(10, "-"))       # --------hi  -> text on the right
print("42".zfill(5))             # 00042 -> pads zeros on the left (good for IDs)
print("-42".zfill(5))            # -0042 -> sign stays in front
print("a\tb\tc".expandtabs(4))   # tab character becomes spaces (tab size 4)
````

## Checking methods : isalpha, isdigit, isalnum and friends
All of these return `True` or `False`. Empty string always gives `False` for them.

````python
print("hello".isalpha())        # True  -> only letters
print("hello1".isalpha())       # False -> digit is there
print("2026".isdigit())         # True  -> only digits
print("2026".isnumeric())       # True  -> numeric characters (also handles things like '½')
print("2026".isdecimal())       # True  -> strictest, only 0-9 type digits
print("abc123".isalnum())       # True  -> letters + digits only
print("abc 123".isalnum())      # False -> space is not allowed
print("   ".isspace())          # True  -> only whitespace
print("HELLO".isupper())        # True  -> all cased letters are capital
print("hello".islower())        # True  -> all cased letters are small
print("Hello World".istitle())  # True  -> each word starts with capital
print("my_var1".isidentifier()) # True  -> valid Python variable name
print("1var".isidentifier())    # False -> variable cannot start with a digit
print("hello".isprintable())    # True  -> no hidden characters like newline
print("hello\n".isprintable())  # False
print("hello".isascii())        # True  -> only plain English keyboard characters
print("नमस्ते".isascii())        # False -> Hindi letters are not ASCII
````

## casefold, format, format_map
- `casefold()` is a stronger `lower()`, mainly for comparing text ignoring case in all languages
- `format()` fills the `{ }` holes in a string, older style than f-strings

````python
print("Straße".lower())         # straße  -> lower() leaves the German ß as it is
print("Straße".casefold())      # strasse -> casefold() converts it, so comparison works
print("HELLO".casefold() == "hello".casefold())  # True

print("My name is {} and I am {}".format("Rohan", 30))          # filled in order
print("My name is {1} and I am {0}".format(30, "Rohan"))       # by position number
print("My name is {n} from {c}".format(n="Rohan", c="Gurugram"))  # by name
print("Pi is {:.2f}".format(3.14159))                          # .2f -> 2 digits after decimal

data = {"n": "Rohan", "c": "Gurugram"}
print("{n} lives in {c}".format_map(data))   # takes values straight from a dictionary
````

## removeprefix and removesuffix (Python 3.9+)
Cuts a piece from the start or end only if it is actually there, otherwise returns the string unchanged.

````python
file = "report_final.pdf"
print(file.removeprefix("report_"))   # final.pdf
print(file.removesuffix(".pdf"))      # report_final
print(file.removesuffix(".txt"))      # report_final.pdf -> no match, nothing removed
````

## translate and maketrans
`maketrans()` makes a replacement table, `translate()` applies it. Faster than many `replace()` calls when changing single characters.

````python
table = str.maketrans("aeiou", "12345")   # a->1, e->2, i->3, o->4, u->5
print("education".translate(table))        # 2d5c1t34n

table2 = str.maketrans("", "", "aeiou")    # third argument = characters to delete
print("education".translate(table2))        # dctn
````

## encode (and decode)
Computers store text as bytes. `encode()` converts a string to bytes, and `decode()` brings it back.

````python
b = "Rohan".encode("utf-8")      # string -> bytes
print(b)                         # b'Rohan'
print(type(b))                   # <class 'bytes'>
print("नमस्ते".encode("utf-8"))   # non-English letters take more than 1 byte each
print(b.decode("utf-8"))         # bytes -> back to string
````

## Operators that work on strings : `in`, `*`, comparison, reverse

````python
print("john" in x)               # True  -> `in` checks if a piece exists
print("xyz" not in x)            # True
print("ha" * 3)                  # hahaha -> repetition using *
print("apple" < "banana")        # True -> compares alphabetically (letter by letter)
print("a" == "A")                # False -> capital and small are different
print(x[::-1])                   # step -1 -> reverses the string
print(x[::2])                    # step 2 -> every second character
print(ord("A"), chr(65))         # 65 A -> character <-> its number code
````
