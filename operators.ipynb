# operators

Operators are the symbols that make python **do something** with values, like `+` to add or `>` to compare. The values they work on are called **operands**. In `12 + 3`, the `+` is the operator and 12 and 3 are the operands.

We have covered four kinds so far:
1. arithmetic -> maths
2. assignment -> storing a value in a variable
3. comparison -> asking a question, answer is `True` or `False`
4. logical -> joining several `True`/`False` answers together

## 1. arithmetic operators

| operator | name | example (a=12, b=3) |
|---|---|---|
| `+` | add | 15 |
| `-` | subtract | 9 |
| `*` | multiply | 36 |
| `/` | divide (**always gives a float**) | 4.0 |
| `%` | modulus, the remainder left after dividing | 0 |
| `**` | power | 1728 |
| `//` | floor division, the quotient with the decimal part cut off | 4 |

```python
# two variables to work with
a = 12
b = 3
# each line below does one calculation and prints the answer
print(a+b) # sum
print(a-b) # difference
print(a*b) # product
print(a/b) # quotient, note the answer is 4.0 and not 4, / always returns a float
print(a%b) # remainder, 12 divides by 3 fully so nothing is left -> 0
print(a**b) # exponential (power), 12 x 12 x 12 = 1728
print(a//b) # floor division (quotient rounded off to the previous whole number)
```

```python
# floor division vs normal division with a number that doesn't divide evenly
print(17 / 5)  # 3.4, the full answer
print(17 // 5) # 3, only the whole part, decimal is thrown away
print(17 % 5)  # 2, what is left over (5 x 3 = 15, and 17 - 15 = 2)
```

> `%` is very handy: `n % 2` is `0` for an even number and `1` for an odd number.

## 2. assignment operators

`=` stores a value in a variable. The others are shortcuts: `x += 10` means `x = x + 10`. The variable is changed using its own old value.

| shortcut | same as |
|---|---|
| `x += 10` | `x = x + 10` |
| `x -= 10` | `x = x - 10` |
| `x *= 10` | `x = x * 10` |
| `x /= 10` | `x = x / 10` |
| `x %= 10` | `x = x % 10` |
| `x **= 10` | `x = x ** 10` |
| `x //= 10` | `x = x // 10` |

```python
x = 23 # regular assignment, x now holds 23

x += 10 # add and assign -> 23 + 10 = 33
print(x)
x -= 10 # subtract and assign -> 33 - 10 = 23 (back to where we started)

print(x)
x *= 10 # multiply and assign -> 23 * 10 = 230

print(x)
x /= 10 # divide and assign -> 230 / 10 = 23.0 (becomes a float because of /)
print(x)

x %= 10 # remainder and assign -> 23.0 % 10 = 3.0
print(x)
x **= 10 # power and assign -> 3.0 to the power 10 = 59049.0
print(x)
x //= 10 # floor division and assign -> 59049.0 // 10 = 5904.0
print(x)
```

## 3. comparison operators

They compare two values and the answer is always a **boolean**: `True` or `False`.

| operator | meaning |
|---|---|
| `>` | greater than |
| `>=` | greater than or equal to |
| `<` | less than |
| `<=` | less than or equal to |
| `==` | equal to |
| `!=` | not equal to |

Careful: a single `=` **stores** a value, a double `==` **asks** if two values are equal.

```python
# a is 12 and b is 3 (from the first cell)
print(a > b)  # is 12 greater than 3 -> True
print(a >= b) # greater than or equal to -> True
print(a < b)  # is 12 less than 3 -> False
print(a <= b) # less than or equal to -> False
print(a == b) # are they equal -> False
print(a != b) # are they NOT equal -> True
```

```python
# comparison works on strings too, == checks if the text is exactly the same
print("python" == "python") # True
print("python" == "Python") # False, capital P makes it different
print("apple" < "banana")   # True, strings are compared alphabetically
```

## 4. logical operators

They join conditions (things that are `True` or `False`).

- `and` -> **all** conditions must be true
- `or` -> **at least one** condition must be true
- `not` -> flips the answer, `True` becomes `False` and `False` becomes `True`

**truth table**

| A | B | A and B | A or B |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

```python
# and : expects all the conditions to be true
print(12 > 10 and 10 < 12) # t and t = t
print(12 < 10 and 10 < 12) # f and t = f
print(12 < 10 and 10 > 12) # f and f = f
```

```python
# or : expects at least one of the conditions to be true
print(12 > 10 or 10 > 12) # t or f = t
print(12 < 10 or 10 < 12) # f or t = t
print(12 < 10 or 10 > 12) # f or f = f, no condition is true
```

```python
# not : reverse boolean
print(not True)    # False
print(not 12 > 10) # 12 > 10 is True, not flips it -> False
```

```python
# real example: a person can vote if age is 18 or more AND they are a citizen
age = 20
is_citizen = True
print(age >= 18 and is_citizen) # both are true -> True
```
